```python
import pandas as pd

url = "https://raw.githubusercontent.com/owid/covid-19-data/master/public/data/vaccinations/vaccinations.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df.head())
```

```python

print("Dataset Information:")
df.info()
```

```python

print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python

print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python

df["date"] = pd.to_datetime(df["date"], errors="coerce")

print("Start Date:", df["date"].min())
print("End Date:", df["date"].max())
```

```python

df.to_csv("cleaned_covid_vaccination_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python

# Get the latest available record for each location
latest = (
    df.dropna(subset=["people_vaccinated_per_hundred"])
      .sort_values("date")
      .groupby("location")
      .tail(1)
)

# Exclude aggregate rows, not individual countries
aggregates = [
    "World", "European Union", "High income",
    "Low income", "Upper middle income",
    "Lower middle income", "International"
]

latest_countries = latest[~latest["location"].isin(aggregates)]

top10 = latest_countries.sort_values(
    "people_vaccinated_per_hundred",
    ascending=False
).head(10)

display(
    top10[["location", "date", "people_vaccinated_per_hundred"]]
)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top10,
    x="people_vaccinated_per_hundred",
    y="location",
    color="royalblue"
)

plt.title("Top 10 Locations by COVID-19 Vaccination Coverage")
plt.xlabel("People Vaccinated per 100 Population (%)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

```python

world = df[df["location"] == "World"].copy()

world = world.dropna(
    subset=["people_vaccinated_per_hundred"]
).sort_values("date")

plt.figure(figsize=(12, 6))

plt.plot(
    world["date"],
    world["people_vaccinated_per_hundred"],
    color="green"
)

plt.title("Global COVID-19 Vaccination Trend")
plt.xlabel("Date")
plt.ylabel("People Vaccinated per 100 Population (%)")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

countries = ["India", "United States", "United Kingdom"]

compare = df[df["location"].isin(countries)].copy()

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=compare,
    x="date",
    y="people_vaccinated_per_hundred",
    hue="location"
)

plt.title("COVID-19 Vaccination Coverage: India vs USA vs UK")
plt.xlabel("Date")
plt.ylabel("People Vaccinated per 100 Population (%)")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
